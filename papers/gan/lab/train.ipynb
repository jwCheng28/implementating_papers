```python
import os
import pytorch_lightning as pl
from pytorch_lightning.loggers import TensorBoardLogger
```

```python
from src.dataset import FashionMNISTDataModule
```

```python
from src.model import GANLightning
```

```python
pl.seed_everything(42)
    
data_module = FashionMNISTDataModule(batch_size=128, num_workers=4)
    
model = GANLightning(latent_dim=100, lr=0.0002)
    
logger = TensorBoardLogger(save_dir="lightning_logs/", name="dcgan_fashion_mnist")
    
trainer = pl.Trainer(
    max_epochs=10,            
    accelerator="auto",       
    devices=1,                
    logger=logger,
    precision="32",         
    deterministic=False       
)
    
print("Starting GAN training")
trainer.fit(model, datamodule=data_module)
print("Training completed")
```
