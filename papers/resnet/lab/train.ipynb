```python
from src.dataset import CIFAR10DataModule
from src.model import resnet50
import pytorch_lightning as pl

pl.seed_everything(42)
    
data_module = CIFAR10DataModule(batch_size=128)

# define hyper_param for resnet50
model = resnet50(num_classes=10, lr=0.1, weight_decay=0.0001)
```

```python
trainer = pl.Trainer(
        max_epochs=20,
        accelerator="auto",
        devices=1,
    )
    
trainer.fit(model, datamodule=data_module)
```
